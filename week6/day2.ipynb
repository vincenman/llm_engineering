```python
MODEL = "openai/gpt-oss-20b"
```

```python
def make_jsonl(item):
    body = {"model": MODEL, "messages": [{"role": "system", "content": SYSTEM_PROMPT}, {"role": "user", "content": item.full}], "reasoning_effort": "low"}
    line = {"custom_id": str(item.id), "method": "POST", "url": "/v1/chat/completions", "body": body}
    return json.dumps(line)
```

```python
from groq import Groq
import time
import json

client = Groq()

def process_jsonl(file_path, model="llama-3.1-8b-instant"):
    results = []
    with open(file_path, "r", encoding="utf-8") as f:
        for i, line in enumerate(f):
            req = json.loads(line)
            # Extract the body that would have been sent in the batch
            body = req.get("body", req)
            
            try:
                response = client.chat.completions.create(
                    model=body.get("model", model),
                    messages=body["messages"],
                    # add any other params you need
                )
                results.append({
                    "custom_id": req.get("custom_id", f"req-{i}"),
                    "response": response
                })
            except Exception as e:
                print(f"Error on line {i}: {e}")
                time.sleep(2)  # simple backoff
                continue
            
            # very light rate limiting
            if i % 10 == 0:
                time.sleep(0.5)
    
    return results

# Usage
results = process_jsonl("jsonl/0_1000.jsonl")
```

```python
import time
import json
from groq import Groq, RateLimitError

client = Groq()

def process_with_backoff(file_path, model="openai/gpt-oss-20b"):
    results = []
    
    with open(file_path, "r", encoding="utf-8") as f:
        lines = f.readlines()
    
    for i, line in enumerate(lines):
        req = json.loads(line)
        body = req.get("body", req)
        
        max_retries = 5
        for attempt in range(max_retries):
            try:
                response = client.chat.completions.create(
                    model=body.get("model", model),
                    messages=body["messages"],
                    # keep max_tokens low if possible to reduce TPM usage
                )
                results.append({
                    "custom_id": req.get("custom_id", f"req-{i}"),
                    "response": response
                })
                print(f"✓ Line {i+1}/{len(lines)}")
                break   # success → next request
                
            except RateLimitError as e:
                wait = 15 + attempt * 5   # start at 15s, increase
                print(f"Rate limit on line {i+1}. Waiting {wait}s... ({e})")
                time.sleep(wait)
            except Exception as e:
                print(f"Error on line {i+1}: {e}")
                break
        
        # Extra safety pause even after success
        time.sleep(3)   # 3–5 seconds is usually enough on free tier
    
    return results
```
